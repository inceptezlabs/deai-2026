##Let us learn first about Magical Commands
Magical commands are used to apply different types of operations in a single notebook like linux, cloud storage (DBFS commands, ) python programming, spark sql querying etc.,... this is called as a concept of POLYGLOT<br>
**Important Magic Commands**
- %md: allows you to write markdown text to design the notebook (2_Explore_Notebook_Markdowns.ipynb)
- %run: runs a Python file or a notebook.
- %sh: executes shell commands on the cluster edge/client node.
<br>/home/irfan/folder1/file.txt - single linux server harddisk</br>
- %fs: allows you to interact with the Databricks file system (Datalake command (cloud storage s3/adls/gcs))
dbfs:/Volumes/workspace/default/datalake1/accounts.csv - cloud storage
- %sql: allows you to run Spark SQL/HQL queries (ISO/Ansi SQL).
- %python: switches the notebook context to Python.
- %pip: allows you to install Python packages.

**Not Important Magic Commands or We learn few of these where we have Cloud(Azure) dependency**
- %scala: switches the notebook context to Scala.
- %r: switches the notebook context to R.
- %lsmagic: lists all the available magic commands.
- %config: allows you to set configuration options for the notebook.
- %load: loads the contents of a file into a cell.
- %who: lists all the variables in the current scope.

This %fs will be performing DBFS backed by Cloud filesystem (Cloud Storage Area Network (Multiple harddisks kept across datacenters))
<br>How to run a DBFS (like Hadoop) FS commands inside a notebook using %fs magic command to copy the uploaded data into some other volume from the uploaded volume
</br> this **%fs ls** is equalent to **hadoop fs -ls ** command we used in hadoop
</br> this **%fs ls** is listing DBFS files/folders which is a wrapper on top of some Cloud Storage (Cloud Datalake)

```python
%fs ls "dbfs:///Volumes/workspace/default/volumewd38_datalake/dbfs_cloudfs_data/"
```

```python
%fs cp "dbfs:/Volumes/workspace/default/volumewd38_datalake/dbfs_cloudfs_data/accounts.csv" "dbfs:/Volumes/workspace/default/volumewd38_datalake/dbfs_cloudfs_data/accounts_backup.csv"
```

Learning for the first time the dbutils, we learn in detail later

```python
%python
dbutils.fs.cp("dbfs:/Volumes/workspace/default/volumewd38_datalake/dbfs_cloudfs_data/accounts.csv","dbfs:/Volumes/workspace/default/volumewd38_datalake/dbfs_cloudfs_data/accounts_backup2.csv")
dbutils.fs.rm("dbfs:/Volumes/workspace/default/volumewd38_datalake/dbfs_cloudfs_data/accounts_backup2.csv")
```
